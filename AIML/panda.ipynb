```python
import pandas as pd
data={'name':['Pavan','kapil','Lalit','Ishan','Om'],'age':[25,None,44,23,None],'salary':[5000,6000,7000,None,None],'city':['Hyd','Banglore','Pune','Chennai','Delhi']}
df=pd.DataFrame(data)
print(df)
print(df.isnull().sum())
df1=df.dropna()
print(df1)
df['age'].fillna(df['age'].mean(),inplace=True)
print(df)
df=pd.read_csv('ID,Name,Gender,Age,Salary,Purchased.csv')
print(df)
from sklearn.preprocessing import LabelEncoder,MinMaxScaler
le=LabelEncoder()
df['Gender_encoded']=le.fit_transform(df['Gender'])
df['Purchased_encoded']=le.fit_transform(df['Purchased'])
print(df)
df=pd.get_dummies(df,columns=['Name'])
print(df)
age_scaler=MinMaxScaler()
salary_scaler=MinMaxScaler()
id_scaler=MinMaxScaler()
df[['ID']]=id_scaler.fit_transform(df[['ID']])
df[['Age']]=age_scaler.fit_transform(df[['Age']])
df[['Salary']]=salary_scaler.fit_transform(df[['Salary']])
print(df)
from sklearn.model_selection import train_test_split
x=df[['Age']]
y=df[['Salary']]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print("training_data:")
age_train=age_scaler.inverse_transform(x_train)
salary_train=salary_scaler.inverse_transform(y_train)
print(age_train)
print(salary_train)
print("test_data:")
age_test=age_scaler.inverse_transform(x_test)
salary_test=salary_scaler.inverse_transform(y_test)
print(age_test)
print(salary_test)
```
