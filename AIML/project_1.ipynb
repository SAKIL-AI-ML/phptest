```python
#decision tree classifier ml_model
import pandas as pd
df=pd.read_csv("details.csv")
from sklearn.preprocessing import StandardScaler,LabelEncoder
le1=LabelEncoder()
le2=LabelEncoder()
df['Gender']=le1.fit_transform(df['Gender'])
df['Purchased']=le2.fit_transform(df['Purchased'])
scaler=StandardScaler()
df[['Age','Salary']]=scaler.fit_transform(df[['Age','Salary']])
print(df)
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
x=df[['Gender','Age','Salary']]
y=df['Purchased']
model.fit(x,y)
name=input("enter the name")
gender=input("enter the gender")
age=int(input("enter the age"))
salary=int(input("enter the salary"))
gender_input_encoded=le1.transform([gender])
scaled_values=scaler.transform([[age,salary]])
age_scaled=scaled_values[0][0]
salary_scaled=scaled_values[0][1]
prediction=model.predict([[gender_input_encoded[0],age_scaled,salary_scaled]])[0]
if(prediction==0):
    print(name,"will not purchase the product")
else:
    print(name,"will purchase the product")









```
