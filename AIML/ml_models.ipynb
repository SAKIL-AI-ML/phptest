```python
#linear regression ml_model
from sklearn.linear_model import LinearRegression
x=[[1],[2],[3],[4]]
y=[10,20,30,40]
model=LinearRegression()
model.fit(x,y)
hour=int(input("enter how many hours u want to study:"))
print("marks will be:",model.predict([[hour]])[0])
#knn ml_model
from sklearn.neighbors import KNeighborsClassifier
x=[[5,1],[6,2],[7,3],[8,4],[9,5]]
#0=fail,1=pass
y=[0,0,1,1,1]
model=KNeighborsClassifier(n_neighbors=3)
model.fit(x,y)
hour=int(input("enter how many hours he studied"))
iq=int(input("enter his iq"))
prediction=model.predict([[hour,iq]])[0]
if(prediction==0):
    print("he will fail")
else:
    print("he will pass")
#decision tree classifier ml_model
import pandas as pd
df=pd.read_csv("details.csv")
from sklearn.preprocessing import StandardScaler,LabelEncoder
le=LabelEncoder()
df['Gender']=le.fit_transform(df['Gender'])
df['Purchased']=le.fit_transform(df['Purchased'])
scaler=StandardScaler()
df[['Age','Salary']]=scaler.fit_transform(df[['Age','Salary']])
print(df)
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
x=df[['Gender','Age','Salary']]
y=df['Purchased']
model.fit(x,y)
name=input("enter the name")
gender=input("enter the gender")
age=int(input("enter the age"))
salary=int(input("enter the salary"))
gender_input_encoded=le.transform([gender])
age_scaled=scaler.transform([[age]])
salary_scaled=scaler.transform([[salary]])
prediction=model.predict([[gender_input_encoded,age_scaled,salary_scaled]])[0]
if(prediction==0):
    print(name,"will not purchase the product")
else:
    print(name,"will purchase the product")








```
