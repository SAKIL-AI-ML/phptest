```python
import pandas as pd
df=pd.read_csv('details.csv')
print(df)
from sklearn.preprocessing import LabelEncoder,StandardScaler,MinMaxScaler
le=LabelEncoder()
df2=df.copy()
df2['label_encoded']=le.fit_transform(df2['Gender'])
print(df2)
df2.rename(columns={'label_encoded':'Gender_encoded'},inplace=True)
print(df2)
df3=df.copy()
df3['purchased_encoded']=le.fit_transform(df3['Purchased'])
print(df3)
df4=pd.get_dummies(df3,columns=['Age'])
print(df4)
df5=df.copy()
scaler=StandardScaler()
df5[['Salary_scaled']]=scaler.fit_transform(df5[['Salary']])
print(df5)
df5[['Age_scaled']] = scaler.fit_transform(df5[['Age']])
print(df5)
df6=pd.DataFrame({'hours':[1,2,3,4,5],'score':[10,20,30,40,50]})
print(df6)
scaler2=MinMaxScaler()
df6[['hours_scaled','score_scaled']]=scaler2.fit_transform(df6[['hours','score']])
print(df6)
from sklearn.model_selection import train_test_split
x=df6[['hours_scaled']]
y=df6[['score_scaled']]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print("training data")
train_scaled=pd.concat([x_train,y_train],axis=1)
print(scaler2.inverse_transform(train_scaled))
print("testing data")
test_scaled=pd.concat([x_test,y_test],axis=1)
print(scaler2.inverse_transform(test_scaled))
```
